```python

import pandas as pd
df=pd.read_csv(r"E:\task_05\dataset\train.csv")
```

```python
df.isnull().sum()
```

```python
df=df.drop(columns="Cabin")
```

```python
df["Age"]=df["Age"].fillna(df["Age"].mean())
```

```python
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df.duplicated().sum()
```

```python
df['Age'] = df['Age'].astype(int)
```
