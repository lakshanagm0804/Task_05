```python
## Visualization##
```

```python
##Univariate Analysis
```

```python
##Survival count
```

```python
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.show()
```

```python
##Observation:
##More passengers died than survived.
```

```python
##Gender Distribution 
```

```python
sns.countplot(x='Sex', data=df)
plt.title("Gender Distribution")
plt.show()
```

```python
##Observation:
##Male passengers were more than female passengers.
```

```python
##Passenger Class
```

```python
sns.countplot(x='Pclass', data=df)
plt.title("Passenger Class Distribution")
plt.show()
```

```python
##Observation:
##Most passengers traveled in third class.
```

```python
## Histogram Graph##
```

```python
##Age Distribution
plt.figure(figsize=(6,4))
plt.hist(df['Age'], bins=20)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
##Observations
##Most passengers were between 20 and 40 years old.
```

```python
##Fare Distribution
plt.figure(figsize=(6,4))
plt.hist(df['Fare'], bins=20)
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")
plt.show()
```

```python
##Observation:
##Most passengers paid lower fares, with a few paying very high fares.
```

```python
##Bivariate Analysis
```

```python
##Survival vs Gender
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Survival by Gender")
plt.show()
```

```python
##Observation:
##Females had a much higher survival rate than males.
```

```python
##Survival vs Passenger Class
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title("Survival by Passenger Class")
plt.show()
```

```python
##Observation:
##First-class passengers survived more often than third-class passengers.
```

```python
##BoxPlots
```

```python
##Age
sns.boxplot(y=df['Age'])
plt.title("Age Boxplot")
plt.show()
```

```python
##Observation:
##The Age column contains a few outliers.
```

```python
##Fare
sns.boxplot(y=df['Fare'])
plt.title("Fare Boxplot")
plt.show()
```

```python
##Observation:
##Several high-value outliers are present in the Fare column.
```

```python
##Survived and Age
sns.boxplot(x='Survived', y='Age', data=df)
plt.show()
```

```python
##Observation:
##Survivors were spread across different age groups, with many younger passengers surviving
```

```python
##Scatterplot
```

```python
##Age vs Fare
plt.figure(figsize=(6,4))
plt.scatter(df['Age'], df['Fare'])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.show()
```

```python
##Observation:
##There is no strong relationship between Age and Fare.
```

```python
##Pairplot
sns.pairplot(df[['Survived','Pclass','Age','SibSp','Parch','Fare']])
plt.show()
```

```python
##Heatmap
plt.figure(figsize=(8,6))
sns.heatmap(
    df[['Survived','Pclass','Age','SibSp','Parch','Fare']].corr(),
    annot=True,
    cmap='coolwarm'
)
plt.show()
```

```python
##Observation:
##Passenger class and fare have a noticeable negative correlation.
```

# Summary of Findings

- The Titanic dataset contains 891 passenger records.
- After data cleaning, no missing values remained.
- Most passengers were between 20 and 40 years old.
- Most passengers traveled in third class.
- Male passengers were more numerous than female passengers.
- Female passengers had a higher survival rate than males.
- First-class passengers had better survival chances than third-class passengers.
- Fare and Passenger Class show a moderate negative correlation.
